# Block 1 — Import and check OpenCV

```python
import cv2
print(cv2.__version__)
#OpenCV is a library for working with images and video. Every frame from your webcam is just a NumPy array of pixel values. That's all it is.
```

# Block 2 — Read one image

```python
import cv2
img = cv2.imread("any_image.jpg")
print(type(img))
print(img.shape)# (height, width, 3)
                # 3 = BGR channels (Blue Green Red)


#Notice OpenCV uses BGR not RGB.
# Blue comes first, not Red. 
# Keep this in mind — MediaPipe needs RGB so we'll convert later.
```

# Block 4 — Open your webcam

```python
import cv2
cap = cv2.VideoCapture(0)# 0 = default webcam
# Check if webcam opened
print("Webcam opened:", cap.isOpened())

# Read ONE frame
ret,frame=cap.read()
# ret   = True if frame was read successfully
# frame = the actual image (numpy array)

print("Frame read:", ret)
print("Frame shape:", frame.shape)  # (height, width, 3)
# Always release when done
cap.release()
```

# Block 5 — Live webcam feed

```python
cap =cv2.VideoCapture(0)

while True:
    ret,frame=cap.read()
    if not ret:
        break
    cv2.imshow("Webcam",frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

# Block 6 — Draw on frames

```python
cap= cv2.VideoCapture(0)
while True:
    ret,frame=cap.read()
    if not ret: 
        break
    h,w=frame.shape[:2]# Get frame dimensions
    # Draw a rectangle
    # cv2.rectangle(image, top-left, bottom-right, color BGR, thickness)
    cv2.rectangle(frame, (50, 50), (200, 150), (0, 255, 0), 2)

    # Draw text
    # cv2.putText(image, text, position, font, size, color, thickness)
    cv2.putText(frame, "Hello!", (50, 40),
    cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)

    # Draw a circle
    center = (w // 2, h // 2)  # middle of frame
    cv2.circle(frame, center, 50, (255, 0, 0), 2) 

    cv2.imshow("Drawing", frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break      

cap.release()
cv2.destroyAllWindows()
```

# Block 7 — Convert BGR to RGB

```python
#MediaPipe needs RGB images. OpenCV gives BGR. You'll do this conversion on every single frame:

cap = cv2.VideoCapture(0)

while True:
    ret,frame=cap.read()
    if not ret:
        break

    # Convert BGR → RGB
    rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    # Show both side by side to see difference
    cv2.imshow("BGR (OpenCV)", frame)
    cv2.imshow("RGB", rgb_frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

# Block 8 — Show text on live feed (like your dashboard)

```python
cap = cv2.VideoCapture(0)

while True:
    ret,frame = cap.read()
    if not ret:
        break

    h,w = frame.shape[:2]
    # Status text top left
    cv2.putText(frame, "Status: Safe", (20, 40),
    cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 0), 2)


    #EAR value
    cv2.putText(frame, "EAR: 0.35", (20, 80),
                cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)

    cv2.imshow("Live Feed", frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break   

cap.release()
cv2.destroyAllWindows()
```
