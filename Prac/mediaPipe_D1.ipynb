# Block 1 — Import and setup MediaPipe

```python
import cv2 
import mediapipe as mp

# MediaPipe has many solutions — we need face mesh
mp_face_mesh=mp.solutions.face_mesh
mp_draw =mp.solutions.drawing_utils

face_mesh=mp_face_mesh.FaceMesh(
    max_num_faces=1,# only track 1 face
    min_detection_confidence=0.5,# how confident before detecting
    min_tracking_confidence=0.5# how confident to keep tracking
)
print("MediaPipe ready")
```

# Block 2 — Understand what landmarks are

```python
# MediaPipe detects 468 points on your face.
# Each point has:
#   .x  → horizontal position (0.0 to 1.0)
#   .y  → vertical position   (0.0 to 1.0)
#   .z  → depth

# They are normalized — meaning 0.0 is left edge,
# 1.0 is right edge of the frame.

# To get pixel coordinates:
#   pixel_x = landmark.x * frame_width
#   pixel_y = landmark.y * frame_height
```

# Block 3 — Detect face and draw all landmarks

```python
mp_face_mesh = mp.solutions.face_mesh
mp_draw      = mp.solutions.drawing_utils

face_mesh = mp_face_mesh.FaceMesh(
    max_num_faces            = 1,
    min_detection_confidence = 0.5,
    min_tracking_confidence  = 0.5
)

cap =cv2.VideoCapture(0)

while True:
    ret,frame=cap.read()
    if not ret:
        break

    h,w =frame.shape[:2]
    # Convert BGR → RGB (MediaPipe needs RGB)
    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    # Process frame — detect face landmarks
    results = face_mesh.process(rgb)

    #Draw landmarks if face detected
    if results.multi_face_landmarks:
        for face_landmarks in results.multi_face_landmarks:
            mp_draw.draw_landmarks(
                frame,# draw on original BGR frame
                face_landmarks,# the 468 points
                mp_face_mesh.FACEMESH_CONTOURS
            )

    cv2.imshow("Face Mesh", frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break


cap.release()
cv2.destroyAllWindows()
```

# Block 4 — Access individual landmark coordinates

```python
import cv2
import mediapipe as mp

mp_face_mesh = mp.solutions.face_mesh
face_mesh    = mp_face_mesh.FaceMesh(max_num_faces=1,
               min_detection_confidence=0.5,
               min_tracking_confidence=0.5)

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    h, w = frame.shape[:2]
    rgb  = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    results = face_mesh.process(rgb)

    if results.multi_face_landmarks:
        # get first face
        landmarks= results.multi_face_landmarks[0].landmark
        # Total number of landmarks
        print(f"Total landmarks: {len(landmarks)}")

        # Access landmark number 1 (nose tip)
        nose = landmarks[1]
        print(f"Nose normalized: x={nose.x:.3f} y={nose.y:.3f}")    

        # Convert to pixel coordinates
        nose_x=int(nose.x * w)
        nose_y=int(nose.y*h)
        print(f"Nose pixel: x={nose_x} y={nose_y}")
        # Draw a red dot on nose tip
        cv2.circle(frame, (nose_x, nose_y), 5, (0, 0, 255), -1)


    cv2.imshow("Nose Point", frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

# Block 5 — Draw specific eye landmarks

```python
import cv2
import mediapipe as mp

mp_face_mesh = mp.solutions.face_mesh
face_mesh    = mp_face_mesh.FaceMesh(max_num_faces=1,
               min_detection_confidence=0.5,
               min_tracking_confidence=0.5)

# There are the 6 landmark points for each eye

LEFT_EYE  = [362, 385, 387, 263, 373, 380]
RIGHT_EYE = [33,  160, 158, 133, 153, 144]
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    h, w = frame.shape[:2]
    rgb  = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    results = face_mesh.process(rgb)
    if results.multi_face_landmarks:
        landmarks = results.multi_face_landmarks[0].landmark

    # Draw left eye points in green
    for idx in LEFT_EYE:
        x=int(landmarks[idx].x * w)
        y=int(landmarks[idx].y * h)
        cv2.circle(frame, (x, y), 3, (0, 255, 0), -1)

    # Draw right eye points in blue
        for idx in RIGHT_EYE:
            x = int(landmarks[idx].x * w)
            y = int(landmarks[idx].y * h)
            cv2.circle(frame, (x, y), 3, (255, 0, 0), -1)

    cv2.imshow("Eye points",frame)
    if cv2.waitKey(1) & 0xFF ==ord('q'):
        break

cap.release()
cap.destroyAllWindows()




```

# Block 6 — Show face detected status on screen

```python
import cv2
import mediapipe as mp

mp_face_mesh = mp.solutions.face_mesh
face_mesh    = mp_face_mesh.FaceMesh(max_num_faces=1,
               min_detection_confidence=0.5,
               min_tracking_confidence=0.5)

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    h, w = frame.shape[:2]
    rgb  = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    results = face_mesh.process(rgb)

    if results.multi_face_landmarks:
        landmarks = results.multi_face_landmarks[0].landmark
        # Green text — face found
        cv2.putText(frame, "Face Detected", (20, 40),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 0), 2)
        cv2.putText(frame, f"Landmarks: {len(landmarks)}", (20, 80),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)
    else:
        # Red text — no face
        cv2.putText(frame, "No Face", (20, 40),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 0, 255), 2)

    cv2.imshow("Detection", frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
# results.multi_face_landmarks     → list of all detected faces
# results.multi_face_landmarks[0]  → first face
# .landmark[i]                     → specific point i
# .x and .y                        → normalized 0.0 to 1.0
# multiply by w and h              → pixel coordinates
```
